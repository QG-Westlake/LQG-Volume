```python
va=Symbol('A',commutative=False)
vb=Symbol('B',commutative=False)
vc=Symbol('C',commutative=False)

def commutator(func1):
    f2=func1.subs({vc*va:va*vc-I*t*vb}).expand().subs({vc*vb:vb*vc+I*t*va}).expand().subs({vb*va:va*vb+I*t*vc}).expand()
    if f2!=func1:
        return commutator(f2)
    else:
        return f2
    
def dvlist(f1):
    dlist=[]
    if f1.func==sympy.core.symbol.Symbol:
        if f1==va:
            dlist.append(1)
        elif f1==vb:
            dlist.append(2)
        elif f1==vc:
            dlist.append(3)
    elif f1.func==sympy.core.power.Pow:
        for j in range(0,f1.args[1]):
            if f1.args[0]==va:
                dlist.append(1)
            elif f1.args[0]==vb:
                dlist.append(2)
            elif f1.args[0]==vc:
                dlist.append(3)
    else:
        for i in f1.args:
            if i.func==sympy.core.power.Pow:
                for j in range(0,i.args[1]):
                    if i.args[0]==va:
                        dlist.append(1)
                    elif i.args[0]==vb:
                        dlist.append(2)
                    elif i.args[0]==vc:
                        dlist.append(3)
            if i==va:
                dlist.append(1)
            elif i==vb:
                dlist.append(2)
            elif i==vc:
                dlist.append(3)
    return dlist

def dvn(f1):
    listn=[1]
    listd=[]
    if f1.func==sympy.core.add.Add:
        ls1=[]
        for i in f1.args:
            if i.func==sympy.core.mul.Mul:
                coef1=1
                if type(i.args[0])==sympy.core.numbers.Integer or type(i.args[0])==sympy.core.numbers.NegativeOne:
                    coef1=coef1*(i.args[0])
                if I in i.args:
                    coef1=coef1*I
                if t in i.args:
                    coef1=coef1*t
                for argi in i.args:
                    if argi.func==sympy.core.power.Pow and t in argi.args:
                        coef1=coef1*argi
                ls1.append(coef1)
            else:
                ls1.append(1)
            listd.append(dvlist(i))
        listn=ls1
    if f1.func==sympy.core.mul.Mul:
        coef1=1
        if type(f1.args[0])==sympy.core.numbers.Integer or type(f1.args[0])==sympy.core.numbers.NegativeOne:
            coef1=coef1*(f1.args[0])
        if I in f1.args:
            coef1=coef1*I
        if t in f1.args:
            coef1=coef1*t
        for argi in f1.args:
            if argi.func==sympy.core.power.Pow and t in argi.args:
                coef1=coef1*argi
        listn=[coef1]
        listd.append(dvlist(f1))
    if f1.func==sympy.core.symbol.Symbol or f1.func==sympy.core.power.Pow:
        listd.append(dvlist(f1))
    return (listn,listd)
```
